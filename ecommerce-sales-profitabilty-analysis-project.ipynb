```python
df['Profit_Margin'] = (df['Profit'] / df['Total_Sales']) * 100

df['Year'] = df['Order_Date'].dt.year

df['Month'] = df['Order_Date'].dt.month

df['Month_Name'] = df['Order_Date'].dt.month_name()

df['Order_Value'] = df['Quantity'] * df['Unit_Price']

df['Gross_Order_Value'] = df['Quantity'] * df['Unit_Price']

df['Discount_Amount'] = df['Gross_Order_Value'] * (df['Discount_Percent'] / 100)

df['Profit_Per_Unit'] = df['Profit'] / df['Quantity']

df['Sales_Per_Unit'] = df['Total_Sales'] / df['Quantity']

df['Quarter'] = df['Order_Date'].dt.quarter
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df.describe()
```

```python
# UNIVARIATE ANALYSIS

# HISTOGRAM

plt.figure(figsize=(8,5))

sns.histplot(
    df['Total_Sales'],
    bins=30,
    kde=True
)

plt.title("Distribution of Total Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
#PROFIT DISTRIBUTION

plt.figure(figsize=(8,5))

sns.boxplot(
    y=df['Profit']
)

plt.title("Profit Distribution")

plt.ylabel("Profit Amount")

plt.show()
```

```python
# DISTRIBUTION OF PROFIT MARGIN

plt.figure(figsize=(8,5))

sns.histplot(
    df['Profit_Margin'],
    bins=30,
    kde=True
)

plt.title("Distribution of Profit Margin")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Frequency")

plt.show()
```

```python
# DISTRIBUTION OF ORDER VALUE

plt.figure(figsize=(8,5))

sns.histplot(
    df['Order_Value'],
    bins=30,
    kde=True
)

plt.title("Distribution of Order Value")
plt.xlabel("Order Value")
plt.ylabel("Frequency")

plt.show()
```

```python
# CUSTOMER SEGMENT DISTRIBUTION

segment_count = df['Customer_Segment'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    segment_count,
    labels=segment_count.index,
    autopct='%1.1f%%'
)

plt.title("Customer Segment Distribution")

plt.show()
```

```python
# NUMBER OF ORDERS BY PRODUCT CATEGORY

category_count = df['Product_Category'].value_counts()

plt.figure(figsize=(10,5))

sns.barplot(
    x=category_count.index,
    y=category_count.values
)

plt.title("Number of Orders by Product Category")

plt.xlabel("Product Category")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

plt.show()
```

```python
# ORDERS BY QUARTER

quarter_count = df['Quarter'].value_counts().sort_index()

quarter_count.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Orders by Quarter")

plt.xlabel("Quarter")
plt.ylabel("Number of Orders")

plt.show()
```

```python
# DISCOUNT AMOUNT DISTRIBUTION

plt.figure(figsize=(8,5))

sns.boxplot(
    y=df['Discount_Amount']
)

plt.title("Discount Amount Distribution")

plt.ylabel("Discount Amount")

plt.show()
```

```python
# BIVARIATE ANALYSIS

# SALES VS PROFIT

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Total_Sales',
    y='Profit',
    hue='Product_Category'
)

plt.title("Relationship Between Sales and Profit")

plt.xlabel("Total Sales")
plt.ylabel("Profit")

plt.show()
```

```python
# IMPACT OF DISCOUNT ON PROFIT

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Discount_Percent',
    y='Profit'
)

plt.title("Impact of Discount on Profit")

plt.xlabel("Discount Percentage")
plt.ylabel("Profit")

plt.show()
```

```python
#  RELATIONSHIP BETWEEN SALES AND PROFIT MARGIN

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Total_Sales',
    y='Profit_Margin'
)

plt.title(
    "Relationship Between Sales and Profit Margin"
)

plt.xlabel("Total Sales")
plt.ylabel("Profit Margin (%)")

plt.show()
```

```python
# SALES BY CUSTOMER SEGMENT

segment_sales = df.groupby(
    'Customer_Segment'
)['Total_Sales'].sum()


plt.figure(figsize=(8,5))

sns.barplot(
    x=segment_sales.index,
    y=segment_sales.values
)

plt.title("Sales by Customer Segment")

plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.show()
```

```python
# PROFIT BY PRODUCT CATEGORY

category_profit = df.groupby(
    'Product_Category'
)['Profit'].sum()


plt.figure(figsize=(10,5))

category_profit.sort_values().plot(
    kind='barh'
)

plt.title("Profit by Product Category")

plt.xlabel("Profit")

plt.show()
```

```python
# MULTIVARIATE ANALYSIS

# CORRELATION HEATMAP

numeric_columns = [
    'Quantity',
    'Unit_Price',
    'Discount_Percent',
    'Total_Sales',
    'Shipping_Cost',
    'Profit'
]
corr_matrix = df[numeric_columns].corr()
plt.figure(figsize=(10,7))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# SALES AND PROFITABILITY BY CUSTOMER SEGMENT - PIVOT TABLE

segment_analysis = df.groupby(
    'Customer_Segment'
)[
[
'Total_Sales',
'Profit',
'Profit_Margin'
]
].mean()


segment_analysis
```

```python
# SALES AND PROFITABILITY BY CUSTOMER SEGMENT - VISUALIZATION

segment_analysis.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title(
"Sales and Profitability by Customer Segment"
)

plt.ylabel("Value")

plt.show()
```

```python
# MONTHLY SALES TREND

monthly_sales = df.groupby(
    df['Order_Date'].dt.month
)['Total_Sales'].sum()


plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title("Monthly Sales Trend")

plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
# SALES PERFORMANCE BY REGION AND PRODUCT CATEGORY - PIVOT TABLE

region_category = pd.pivot_table(
    df,
    values='Total_Sales',
    index='Region',
    columns='Product_Category',
    aggfunc='sum'
)

region_category
```

```python
 # SALES PERFORMANCE BY REGION AND PRODUCT CATEGORY - VISUALIZATION

plt.figure(figsize=(10,6))

sns.heatmap(
    region_category,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title(
    "Sales Performance by Region and Product Category"
)

plt.xlabel("Product Category")
plt.ylabel("Region")

plt.show()
```

```python
# QUARTERLY SALES AND PROFIT TREND

quarter_analysis = df.groupby(
'Quarter'
)[
[
'Total_Sales',
'Profit'
]
].sum()


quarter_analysis.plot(
kind='line',
marker='o',
figsize=(8,5)
)

plt.title(
"Quarterly Sales and Profit Trend"
)

plt.xlabel("Quarter")

plt.ylabel("Amount")

plt.show()
```
